## Contexte

Un opérateur télécom veut identifier les clients susceptibles de résilier leur abonnement dans les 3 prochains mois. On explore ici les variables les plus discriminantes avant de construire un modèle prédictif.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Données synthétiques — remplacer par votre dataset réel
data = {
    "segment": ["<1 an", "1–2 ans", "2–3 ans", "3–5 ans", ">5 ans"],
    "taux_churn": [0.32, 0.21, 0.14, 0.08, 0.04],
}
df = pd.DataFrame(data)
```

```python
fig, ax = plt.subplots(figsize=(8, 4))
ax.bar(df["segment"], df["taux_churn"], color="#2563eb")
ax.set_xlabel("Ancienneté client")
ax.set_ylabel("Taux de churn")
ax.set_title("Taux de churn par ancienneté")
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda y, _: f"{y:.0%}"))
plt.tight_layout()
plt.show()
```

## Observation

Le taux de churn est **8× plus élevé** chez les clients de moins d'un an que chez ceux fidélisés depuis plus de 5 ans. L'ancienneté est le premier signal d'alerte — le modèle prédictif doit en faire une variable centrale.
